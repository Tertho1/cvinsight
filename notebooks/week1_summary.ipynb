```python
import pandas as pd
import os

datasets = {
    'NER Resumes'        : 'data/processed/ner_resumes_clean.csv',
    'Structured Resumes' : 'data/processed/structured_resumes_clean.csv',
    'Classification'     : 'data/processed/classification_clean.csv',
    'ATS Scores'         : 'data/processed/ats_scores_clean.csv',
    'Netsol Scores'      : 'data/processed/netsol_clean.csv',
}

print(f"{'Dataset':<25} {'Rows':>8} {'Columns':>10} {'Size (KB)':>12}")
print("─" * 60)

total_rows = 0
for name, path in datasets.items():
    df = pd.read_csv(path)
    size_kb = os.path.getsize(path) / 1024
    print(f"{name:<25} {len(df):>8,} {len(df.columns):>10} {size_kb:>11.1f}")
    total_rows += len(df)

print("─" * 60)
print(f"{'TOTAL':<25} {total_rows:>8,}")
print("\n✓ All 5 processed datasets load without errors")
```
